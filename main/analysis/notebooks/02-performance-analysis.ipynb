```python
import pandas as pd
import matplotlib.pyplot as plt
import json
import glob

results = {}
for exp_path in glob.glob('../results/*/'):
    method = exp_path.split('/')[-2]
    with open(exp_path + 'latency_p95.json') as f:
        latency = json.load(f)
    alloc = pd.read_csv(exp_path + 'allocation.csv')
    results[method] = {'latency': latency, 'alloc': alloc}
```

```python
#Cell 2: Biểu đồ so sánh latency P95


methods = list(results.keys())
p95 = [results[m]['latency']['value'] for m in methods]
plt.bar(methods, p95)
plt.ylabel('Latency P95 (ms)')
plt.title('API Response Time (P95) by Method')
plt.show()
```

```python
#Cell 3: Biểu đồ tỷ lệ đăng ký thành công
success_rates = {}
for m in methods:
    df = results[m]['alloc']
    rate = len(df[df['status'] == 'ALLOCATED']) / len(df) * 100
    success_rates[m] = rate
plt.bar(success_rates.keys(), success_rates.values())
plt.ylabel('Success Rate (%)')
plt.title('Registration Success Rate by Method')
plt.show()
```
